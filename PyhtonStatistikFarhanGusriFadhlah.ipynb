```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.hist(
    df["Final_Score"],
    bins=8,
    edgecolor="black"
)

plt.title("Distribusi Final Score Mahasiswa")
plt.xlabel("Final Score")
plt.ylabel("Frekuensi")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    df["Name"].astype(str),  # Diubah ke string agar tidak error
    df["Final_Score"]
)

plt.title("Final Score Setiap Mahasiswa")
plt.xlabel("Nama Mahasiswa")
plt.ylabel("Final Score")

plt.xticks(rotation=90)
plt.show()
```

```python
nilai = df[["Quiz", "Final_Term"]].mean()

plt.figure(figsize=(8, 6))

plt.bar(
    nilai.index,
    nilai.values
)

plt.title("Rata-rata Nilai Quiz, Mid Term, dan Final Term")
plt.xlabel("Jenis Penilaian")
plt.ylabel("Nilai Rata-rata")

plt.show()
```

```python
grade_count = df["Grade"].value_counts()

plt.figure(figsize=(8, 6))

plt.bar(
    grade_count.index.astype(str),
    grade_count.values
)

plt.title("Distribusi Grade Mahasiswa")
plt.xlabel("Grade")
plt.ylabel("Jumlah Mahasiswa")

plt.show()
```

```python
grade_count = df["Grade"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    grade_count.values,
    labels=grade_count.index.astype(str),
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Persentase Grade Mahasiswa")

plt.show()
```

```python
nilai = df[["Quiz", "Final_Term", "Final_Score"]]

korelasi = nilai.corr()

display(korelasi)

plt.figure(figsize=(8, 6))

sns.heatmap(
    korelasi,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Korelasi Antar Nilai")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

df[["Quiz", "Mid_Term", "Final_Term", "Final_Score"]].boxplot()

plt.title("Boxplot Nilai Mahasiswa")
plt.ylabel("Nilai")

plt.show()
```

```python
statistik = pd.DataFrame({
    "Mean": df[["Quiz", "Final_Term", "Final_Score"]].mean(),
    "Median": df[["Quiz",  "Final_Term", "Final_Score"]].median(),
    "Std Deviasi": df[["Quiz",  "Final_Term", "Final_Score"]].std(),
    "Minimum": df[["Quiz",  "Final_Term", "Final_Score"]].min(),
    "Maximum": df[["Quiz",  "Final_Term", "Final_Score"]].max()
})

display(statistik)
```
